### Sélection des variables pertinentes et ingénierie des variables

```python
#création d'un nouveau dataset
df_1= df.drop(columns=['class','alive','alone','sex','adult_male','who','embark_town'])
df_1
```

```python
#création d'une nouvelle variable résumant de manière pertinente les variables "who" et "sex"
##Définition d'une fonction: on utilise def
def creation(col):
    if col['who'] in ['man','woman']:
        return col['who']
    elif col['who'] in ['child']:
        return 'girl' if col['sex']== 'female' else  'boy'
    else: 
        return None 
df_1['categorie']= df.apply(creation,axis=1)
```

### Encodage (mapping) des variables catégorielles

```python
# fonction pour mapping
def mapping(data=df, legende = {}):
    cat_data = data.select_dtypes(include=['object', 'category']).columns
    for variable in cat_data:
        temp = data[variable].dropna().unique()
        mapper = {temp[i] : i for i in range(0,len(temp))}
        legende[variable] = mapper
        data[variable] = data[variable].dropna().map(mapper)
    return data, legende

#Appel de la fonction mapping
df1 = df_1.copy()
df1, legende = mapping(df1)
df1
```

```python
# Affichage de la légende du mapping
legende
```

### Analyse et gestion des valeurs manquantes

```python
#Valeur manquante
df1.isnull().sum()
```

```python
#Imputation par la médiane
imputer= SimpleImputer(strategy='median')
df2= imputer.fit_transform(df1)
df2= pd.DataFrame(df2,columns=df1.columns)
```

## Apprendre à se connecter à une base de donnée via `SQLAlchemy`

### Importation des dépendances

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sqlalchemy import create_engine
```

### Connexion à la base de données MySQL

```python
data = create_engine('mysql+pymysql://root:@localhost/market_db')
```

### Sélection de la table souhaité et visualisation du contenu

```python
df = pd.read_sql('SELECT * FROM prices_aapl', data).drop(columns=['id'])
```

```python
df.head(25)
```

### Petite visualisation

```python
def scatter_plot(data, x, y, xlabel='X', ylabel='Y', title='Nuage de points'):
    sns.scatterplot(x=x,y=y,data=data)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.title(title)
    plt.show()
```

```python
scatter_plot(df, 'open', 'close')
```

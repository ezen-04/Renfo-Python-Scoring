### Importation des différentes librairies et bibliothèques

Dans tout notre travail, nous auront le plus souvent affaire aux librairies et bibliothèques suivantes (dans un premier temps):
- **Pandas**: Bibliothèque pour la manipulation et l’analyse de données sous forme de tableaux (DataFrame).
    - **Créer un DataFrame** :
        ```python
        import pandas as pd

        data = {"Nom": ["Alice", "Bob"], "Age": [25, 30]}
        df = pd.DataFrame(data)
        print(df)
        ```
    - **Fonctions courantes** :
        - `df.head()` → affiche les premières lignes  
        - `df.describe()` → statistiques descriptives  
        - `df.groupby("col")` → regrouper par colonne  
        - `df.merge()` → fusionner deux DataFrames  

---

- **Numpy**: Bibliothèque pour le **calcul numérique** et la manipulation de tableaux multidimensionnels.

    - **Créer un tableau** :
        ```python
        import numpy as np

        arr = np.array([1, 2, 3, 4])
        print(arr.mean())  # moyenne
        ```
    - **Fonctions courantes** :
        - `np.mean(arr)` → moyenne  
        - `np.std(arr)` → écart-type  
        - `np.dot(a, b)` → produit scalaire  
        - `np.linspace(0, 1, 5)` → valeurs espacées  

---

- **Matplotlib.pyplot**: Bibliothèque de **visualisation** basique.

    - **Tracer un graphique simple** :
        ```python
        import matplotlib.pyplot as plt

        x = [1, 2, 3]
        y = [2, 4, 6]
        plt.plot(x, y, label="Ligne")
        plt.xlabel("X")
        plt.ylabel("Y")
        plt.legend()
        plt.show()
        ```
    - **Fonctions courantes** :
        - `plt.plot()` → courbes  
        - `plt.bar()` → barres  
        - `plt.hist()` → histogrammes  
        - `plt.scatter()` → nuages de points  

---

- **Seaborn**: Bibliothèque de visualisation **statistique** basée sur Matplotlib.

    - **Exemple** :
        ```python
        import seaborn as sns
        import pandas as pd

        df = pd.DataFrame({"Sexe": ["H", "F", "H", "F"], "Age": [20, 22, 25, 30]})
        sns.barplot(x="Sexe", y="Age", data=df)
        ```
    - **Fonctions courantes** :
        - `sns.barplot()` → barres avec moyenne  
        - `sns.histplot()` → histogrammes  
        - `sns.heatmap()` → cartes de chaleur  
        - `sns.pairplot()` → relations entre variables  

---

- **Scipy**: Bibliothèque pour les **calculs scientifiques** (optimisation, statistiques, intégration).

    - **Exemple : calcul statistique** :
        ```python
        from scipy import stats

        data = [1, 2, 3, 4, 5]
        print(stats.ttest_1samp(data, 3))  # test t
        ```
    - **Fonctions courantes** :
        - `stats.ttest_ind(a, b)` → test t entre deux échantillons  
        - `stats.norm.pdf(x)` → densité loi normale  
        - `integrate.quad()` → intégration numérique  
        - `optimize.minimize()` → optimisation  

---

- **Sqlalchemy**: Bibliothèque pour interagir avec des **bases de données SQL** en Python.

    - **Connexion et requête** :
        ```python
        from sqlalchemy import create_engine

        engine = create_engine("sqlite:///test.db")
        with engine.connect() as conn:
            result = conn.execute("SELECT 1")
            print(result.fetchall())
        ```
    - **Fonctions courantes** :
        - `create_engine()` → créer une connexion  
        - `session.query()` → requêtes ORM  
        - `Table()` et `Column()` → définir des tables  
        - `conn.execute()` → exécuter SQL brut  

---

- **Sklearn**: Bibliothèque pour le **machine learning**.

    - **Exemple : régression linéaire** :
        ```python
        from sklearn.linear_model import LinearRegression
        import numpy as np

        X = np.array([[1], [2], [3]])
        y = np.array([2, 4, 6])

        model = LinearRegression()
        model.fit(X, y)
        print(model.predict([[4]]))  # prédiction
        ```
    - **Fonctions courantes** :
        - `train_test_split()` → séparer données  
        - `StandardScaler()` → normaliser  
        - `LinearRegression()` → régression  
        - `RandomForestClassifier()` → classification  
        - `KMeans()` → clustering
